```python

import pandas as pd

# 1. Charger la base de données
df = pd.read_csv("cupid_algorithm_dataset.csv")

# 2. Créer une nouvelle base
new_df = pd.DataFrame()

# différences A - B
variables = ["age","education","career_ambition","openness","extraversion","agreeableness","conscientiousness","chronotype","spontaneity","emotional_expressiveness"]

for variable in variables:
    new_df[variable + "_difference"] = (
        df["a_" + variable] - df["b_" + variable]
    )

    # Différence en valeur absolue
    new_df[variable + "_absolute_difference"] = (
        new_df[variable + "_difference"].abs()
    )

# 4. Comparer les variables catégorielles soit 1( le meme ) soit 0 different 
categorical_variables = [
    "location",
    "career_field",
    "love_language"
]

for variable in categorical_variables:
    new_df["same_" + variable] = (
        df["a_" + variable] == df["b_" + variable]
    ).astype(int)

# 5. Ajouter la cible à prédire
new_df["compatible"] = df["compatible"]

# 6. Enregistrer la nouvelle base
new_df.to_csv("cupid_differences_A_B.csv", index=False)

# 7. Vérifier le résultat
print("Nouvelle base créée !")
print('dimention:',df.shape)
print("Dimensions :", new_df.shape)

print(new_df.head())
```
